```python
import numpy as np
import matplotlib.pyplot as plt

print("Portfolio Risk Classifier")
print("Project started successfully!")
```

```python
# Covariance matrix
Sigma = np.array([
    [0.04, 0.01],
    [0.01, 0.09]
])

print("Covariance Matrix:")
print(Sigma)
```

```python
w1 = float(input("Enter weight of Asset 1 (0 to 1): "))
w2 = float(input("Enter weight of Asset 2 (0 to 1): "))

if w1 < 0 or w2 < 0 or w1 > 1 or w2 > 1:
    print("Error: weights must be between 0 and 1.")

elif abs(w1 + w2 - 1) > 0.0001:
    print("Error: the two weights must add up to 1.")

else:
    w = np.array([w1, w2])

    print("\nValid portfolio!")
    print("Asset 1 weight:", w1)
    print("Asset 2 weight:", w2)
```

```python
# Quadratic form: w^T Sigma w
variance = w.T @ Sigma @ w

# Portfolio risk
risk = np.sqrt(variance)

print("Portfolio variance =", variance)
print("Portfolio risk =", risk)
```

```python
# Rayleigh quotient
rayleigh = (w.T @ Sigma @ w) / (w.T @ w)

print("Rayleigh quotient =", rayleigh)
```

```python
# Find eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(Sigma)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

print("\nSmallest risk direction:")
print(eigenvectors[:, 0])

print("\nLargest risk direction:")
print(eigenvectors[:, 1])
```

```python
# Create many possible portfolio weights
weights = np.linspace(0, 1, 101)

risks = []

for x in weights:
    portfolio = np.array([x, 1 - x])
    variance = portfolio.T @ Sigma @ portfolio
    risk = np.sqrt(variance)
    risks.append(risk)

# Plot
plt.figure(figsize=(8, 5))

plt.plot(weights, risks)

plt.scatter(w1, risk, s=100)

plt.xlabel("Weight of Asset 1")
plt.ylabel("Portfolio Risk")
plt.title("Portfolio Risk vs Asset 1 Weight")

plt.grid(True)
plt.show()
```
